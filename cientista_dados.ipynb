```python
df_vendas = pd.read_csv("vendas_technordeste.csv")
#shape mostra a composição matricial da do dataframe
print(f'shape: {df_vendas.shape}')
#head mostra as primeiras linhas do dataframe
print(f'head: {df_vendas.head(5)}')
```

```python
#describe mostra algumas funções padrões do pandas aplicadas no dataframe
print(f'describe: {df_vendas.describe()}')
```

```python
#maximo valor na linha em preco_unitario
print(f'maior preço: {df_vendas["preco_unitario"].max()}')
#soma de todas as linhas em quantidade
print(f'produtos vendidos: {df_vendas["quantidade"].sum()}')
```

```python
#media e mediana de desconto_pct
print(f'media: {df_vendas["desconto_pct"].mean()}')
print(f'mediana: {df_vendas["desconto_pct"].median()}')
```

```python
df_vendas.head()
```

```python
#aplica as funções [min max e mean] na coluna frete
df_vendas.agg({
    'frete':["min", "max", "mean"]
})
```

```python
#cria a coluna total_bruto
df_vendas["total_bruto"]=df_vendas["preco_unitario"]*df_vendas["quantidade"]
df_vendas["total_bruto"].head()
```

```python
#cria a coluna total_liquido
df_vendas["total_liquido"]=df_vendas["total_bruto"]*(1-df_vendas["desconto_pct"]/100)+df_vendas["frete"]
df_vendas["total_liquido"].head()
```

```python
#cria a coluna alto_faturamento [True ou False] se total_liquido>3000
df_vendas["alto_faturamento"]=df_vendas["total_liquido"]>3000
df_vendas["alto_faturamento"].head()
```

```python
#cria a coluna avaliado [True ou False] usando avaliacao.notna()
df_vendas["avaliado"] = df_vendas["avaliacao"].notna()
df_vendas["avaliado"].head()
```

```python
#pega da linha 10 a 15 com loc[10:15,[coluna, coluna, coluna]] 
df_vendas.loc[10:15,["produto", "categoria", "cidade"]]
```

```python
#itera o range 15 com i e usa o iloc[linha,coluna] para localizar os precos e os produtos 
for i in range(15):
    preco = df_vendas.iloc[i,5]
    produto = df_vendas.iloc[i,1]
    if preco>=1000:
        print(f"Linha  {i}: produto premium  {produto}  -  R${preco}")
    else:
        print(f"Linha {i}: produto convencional {produto} - R$ {preco}")
```

```python
#define variavel cidade_alvo com Natal Recife
cidade_alvo = ["Natal","Recife"]
#pega cidade que esta em cidade_alvo para isso usa dataframe["coluna"].isin(variavel)
df_filtrado = df_vendas["cidade"].isin(cidade_alvo)
df_filtrado.head()
```

```python
#filtra a categoria notebooks onde quantidade >= 2 
filtrado = df_vendas[
    (df_vendas["categoria"]=="Notebooks")&(df_vendas["quantidade"]>=2)
]
filtrado.head(5)
```

```python
#mostra a todos os items cuja avaliacao >=4.5 classificados por produto cidade avaliacao
df_vendas.loc[df_vendas["avaliacao"]>=4.5,["produto","cidade","avaliacao"]]
```

```python
#mostra por categoria a media do preco_unitario 
receita_final = df_vendas.groupby("categoria")["preco_unitario"].mean()
print(receita_final)
```

```python
#mostra por cidade e categoria a quantidade 
agrupa_cidade_categoria = df_vendas.groupby(["cidade","categoria"])["quantidade"].sum()
print(agrupa_cidade_categoria)
```

```python
#cria a tabela
comparacao = df_vendas.groupby('cidade')['avaliacao'].agg(['size', 'count'])
#adiciona a coluna
comparacao['nao_avaliados'] = comparacao['size'] - comparacao['count']
print(comparacao)
```
